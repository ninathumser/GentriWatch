```python
import joblib
import pandas as pd
import numpy as np
```

```python
import plotly.express as px
import plotly.graph_objects as go
```

```python
import json

with open('./assets/map_pr_new.geojson') as f:
    geomap_pr = json.load(f)
```

```python
df_pr = joblib.load('./dataframes/data_app/final_pr.csv')
```

```python
fig = px.choropleth_mapbox(df_pr,
                           geojson=geomap_pr,
                           featureidkey='properties.SCHLUESSEL',
                           locations='PR',
                           center={'lat': 52.520008, 'lon': 13.404954},
                           zoom=9,
                           color='millenials_perc',
                           color_continuous_scale="Viridis",
                           #range_color=(0, 17),
                           opacity=0.7,
                           animation_frame="year",
                           hover_name='Kiezname',
                           
                           )


fig.update_layout(
    mapbox_style='open-street-map', #"open-street-map", "carto-positron", "carto-darkmatter", "stamen-terrain", "stamen-toner", "stamen-watercolor"
    #mapbox_style="dark",
    height=750,
    margin={"r":0,"t":0,"l":0,"b":0})

fig.show()
```
